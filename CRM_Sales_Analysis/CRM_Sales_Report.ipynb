# Improving CRM Opportunity Conversion

### **Business Analysis Case Study**

> **Business question:** Which opportunity dimensions should sales management focus on to improve conversion and better prioritize sales efforts?

# 1. DEFINE

### Business Situation

Sales management is seeing a large number of opportunities in the CRM, but doesn't have a clear understanding of what is driving successful versus unsuccessful opportunities.

### Objective

Identify the dimensions associated with differences in opportunity conversion and determine where sales management should focus its improvement efforts.

### KPI Baseline

Win Rate = Won Opportunities / (Won Opportunities + Lost Opportunities)

Win rate measures the proportion of finalized opportunities that are successfully converted into wins. In-process opportunities are excluded because they have not yet reached a final outcome and therefore cannot be classified as won or lost.

A higher win rate indicates stronger conversion performance. However, win rates should be interpreted alongside opportunity volume, since small sample sizes can produce unstable or misleading results.

### Target Direction 

Identify opportunity segments with meaningfully lower win rates and sufficient opportunity volume, then determine whether these represent actionable areas for improving conversion and prioritizing sales effort.

# 2. STRUCTURE

To identify potential drivers of opportunity conversion, opportunities were examined across four dimensions:

### Potential Drivers of Win Rate

These dimensions provide the structure for comparing win rates and identifying segments with meaningful differences in conversion performance.

| Deal stage  | Opportunity count |
| ----------- | ----------------: |
| Won         |             4,238 |

Only **Won** and **Lost** opportunities represent finalized outcomes. Prospecting and Engaging opportunities were therefore excluded from win-rate calculations because their outcomes had not yet been determined.

## 6.2 Overall Win Rate

I define win rate as:

**Win Rate = Won Opportunities / (Won + Lost Opportunities)**

Using only finalized opportunities, the overall win rate is **63.15%**.

In other words, approximately **63 out of every 100 finalized opportunities are won**.

This **63.15% benchmark** provides the baseline against which differences across products, sales teams, customers, sales cycles, and timing can be evaluated.

# 7. TEST HYPOTHESES

The initial analysis tested the hypotheses identified in the prioritization stage. SQL was used to compare win rates across key dimensions and identify potentially meaningful patterns. Python was then used to visualize these patterns and investigate the most relevant relationships in greater depth.



## 7.1 Product

### Hypothesis

**Product series have different win rates.**

### Initial analysis

Win rates were compared across product series using finalized opportunities only (`Won` and `Lost`).

### Finding

Win rates across products were relatively similar, with a maximum difference of approximately **4.84 percentage points**.

### Conclusion

The analysis provides limited evidence that product series are a major driver of opportunity conversion.

**Decision:** Move on to other potential drivers.



## 7.2 Sales Organization

### Hypothesis

**Sales teams and sales agents have different win rates.**

### Initial analysis

Win rates were compared across sales agents and sales regions.

### Finding

Sales-agent win rates showed some variation, with a range of **15.41 percentage points**. However, the values were relatively clustered, with a standard deviation of **3.67 percentage points** around an average win rate of **63.51%**.

Sales regions also showed relatively stable win rates.

### Conclusion

The observed variation does not provide strong evidence that sales-agent or regional sales differences are major drivers of conversion.

**Decision:** Move on.



## 7.3 Sales Cycle

### Hypothesis

**Sales-cycle length is associated with opportunity win rates.**

### Initial analysis

Opportunities were grouped into five sales-cycle categories based on the number of days between `engage_date` and `close_date`:

* Very short: 0–30 days
* Short: 31–60 days
* Medium: 61–90 days
* Long: 91–180 days
* Very long: >180 days

### Finding

Longer-cycle opportunities generally showed higher win rates than shorter-cycle opportunities.

The difference between the very-short and long categories was **13.68 percentage points**:

> 57.45% vs. 71.13%

This suggested a potentially meaningful relationship between sales-cycle length and conversion.

### Volume check

The pattern was not driven solely by small sample sizes. The very-short category contained approximately **3,163 opportunities**, while the long category contained approximately **1,358 opportunities**.

### Initial conclusion

Sales-cycle length appears to be associated with win rate, making it an important relationship to investigate further.

However, association alone does not establish that longer sales cycles cause higher conversion.





# 8. CHECK ROBUSTNESS

The sales-cycle finding raised an important question:

> **Could another factor explain the observed relationship?**

Two potential explanations were investigated: opportunity volume and timing.


## 8.1 Opportunity Volume

### Question

**Could high opportunity volume be associated with lower win rates and therefore explain the sales-cycle relationship?**

### Analysis

Opportunity volume was compared with win rates across:

* Sales agents
* Products
* Accounts
* Quarters

### Finding

Higher opportunity volume did not consistently correspond to either higher or lower win rates.

For example, high-volume sales agents and products did not systematically show lower conversion rates. Quarterly volume also did not provide a consistent explanation for differences in win rates.

### Conclusion

Opportunity volume does not provide a clear alternative explanation for the observed relationship between sales-cycle length and win rate.


## 8.2 Sales Cycle × Timing

The next question was whether the relationship between sales-cycle length and win rate remained consistent across quarters.

### Finding

The relationship was **not consistent across time**.

Longer-cycle opportunities generally showed higher win rates from Q2 onward, but this pattern did not hold in Q1. At the same time, Q1 showed an unusually high overall win rate.

### Implication

This suggests that **timing may influence the relationship between sales-cycle length and conversion**.

Therefore, sales-cycle length should not be interpreted as an independent or universal driver of win rate.


# 9. DRILL-DOWN

The interaction between sales-cycle length and timing warranted further investigation.

## 9.1 Quarterly Performance

### Hypothesis

**Opportunity win rates vary across sales periods or seasons.**

### Finding

Q1 exhibited a substantially higher win rate, while Q2–Q4 were considerably lower and relatively stable.

The initial analysis therefore identified **Q1 as a potential point of interest**, rather than a consistent seasonal pattern across all four quarters.

### Business question raised

The sharp difference between Q1 and the remaining quarters could potentially reflect differences in:

* Pipeline composition
* Opportunity quality
* Sales capacity
* Customer purchasing cycles
* Lead generation or qualification
* Other time-related factors

These explanations cannot be established from the available data alone and should therefore be treated as areas for further investigation.


## 9.2 Customer Geography

### Hypothesis

**Customer regions have different win rates.**

### Finding

Customer office locations showed more noticeable variation than sales regions, with a maximum win-rate gap of approximately **18.9 percentage points**.

### Further check

Customer location was compared with customer revenue to investigate whether higher win-rate locations also represented higher-value customers.

The relationship was not consistent: locations with higher win rates did not systematically have higher customer revenue.

### Conclusion

Customer geography may provide useful contextual information, but the observed differences cannot simply be explained by customer revenue.


## 9.3 Customer Industry

### Hypothesis

**Customer industry sectors have different win rates.**

### Finding

Win rates differed by only approximately **3.68 percentage points** across customer sectors.

### Conclusion

Customer industry showed relatively limited variation and does not appear to be a major driver of conversion in this dataset.


## 9.4 Customer Revenue

### Hypothesis

**Customer revenue levels are associated with different win rates.**

### Finding

Win rates differed by approximately **21.87 percentage points** across customer revenue categories.

However, higher-revenue customers did **not** consistently have higher win rates.

### Conclusion

Customer revenue appears to be associated with meaningful differences in conversion, but the relationship is not simply:

> **Higher customer revenue → higher win rate**

This makes customer revenue worth considering as a contextual factor rather than using it as a standalone prioritization rule.


# 10. SYNTHESIZE

The analysis produced several different levels of evidence.

### Stronger findings

**1. Timing is an important contextual factor.**

Q1 showed substantially higher conversion than the remaining quarters, suggesting that opportunity timing deserves further investigation.

**2. Sales-cycle length is associated with conversion, but the relationship is context-dependent.**

Longer-cycle opportunities generally had higher win rates, but this relationship changed across quarters. The interaction analysis therefore provides stronger evidence that **sales-cycle length and timing should be considered together**.

**3. Customer geography shows meaningful variation.**

Customer office locations showed a relatively large difference in win rates, although the available data does not establish why these differences occur.

### Weaker findings

Product series, sales agents, sales regions, and customer industries showed relatively limited or inconsistent differences in win rates.

Customer revenue showed meaningful variation, but higher customer revenue did not consistently correspond to higher conversion.

### Overall interpretation

The analysis suggests that **conversion performance is influenced more by opportunity context and timing than by simply which product, sales agent, or customer industry is involved**.

Most importantly, the relationship between sales-cycle length and conversion appears to be **context-dependent rather than universal**.

# 11. RECOMMEND

Based on the analysis, I would recommend three areas of action.

### 1. Investigate the Q1 → Q2 conversion gap

The sharp difference in conversion between Q1 and the remaining quarters warrants further investigation.

The sales team could examine whether the difference is related to:

* Lead quality
* Pipeline composition
* Sales capacity
* Customer purchasing cycles
* Marketing activity
* Opportunity qualification

### 2. Avoid using sales-cycle length as a standalone prioritization rule

Although longer-cycle opportunities showed higher win rates overall, the relationship changed across quarters.

Sales-cycle length should therefore be combined with other contextual indicators rather than treated as a simple rule such as:

> "Longer cycle = higher-quality opportunity."

### 3. Use multiple signals for opportunity prioritization

Opportunity prioritization could combine:

* Timing
* Sales-cycle characteristics
* Customer context
* Geographic patterns
* Historical conversion patterns

This would provide a more robust basis for deciding where sales resources should be allocated.

# 12. COMMUNICATE

## Executive Insight

**What appears to influence opportunity conversion?**

The analysis suggests that **timing and opportunity context are more informative than product or sales-agent differences alone**.

Sales-cycle length was associated with win rate, but the relationship changed across quarters. Very-short opportunities performed particularly differently outside Q1, suggesting that sales-cycle length should be interpreted in context rather than used as a standalone indicator.

Customer geography and revenue also showed meaningful variation, although the available data does not establish the underlying causes.

**Business implication:** Rather than focusing primarily on product or sales-agent differences, the sales team should investigate the sharp change in conversion after Q1 and develop a more context-aware approach to opportunity prioritization.
